```python
def apply_adaptive_threshold(img):
    img = cv2.GaussianBlur(img, (5, 5), 0)
    img = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 21, 10)
    return cv2.bitwise_not(img)


def scale_image(img):
    return img.astype(np.float32) / 255.0


def reshape_image(img, target_dims=(64, 200)):
    if len(img.shape) == 3 and img.shape[2] == 3:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    height, width = img.shape
    if height > target_dims[0] or width > target_dims[1]:
        scale_factor = min(target_dims[0] / height, target_dims[1] / width)
        img = cv2.resize(img, None, fx=scale_factor, fy=scale_factor, interpolation=cv2.INTER_AREA)

    pad_h = target_dims[0] - img.shape[0]
    pad_w = target_dims[1] - img.shape[1]

    top_pad, bottom_pad = math.ceil(pad_h / 2), math.floor(pad_h / 2)
    left_pad, right_pad = math.ceil(pad_w / 2), math.floor(pad_w / 2)

    img = cv2.copyMakeBorder(img, top_pad, bottom_pad, left_pad, right_pad, cv2.BORDER_CONSTANT, value=255)
    img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    return img


def transform_image(img):
    img = reshape_image(img)
    img = apply_adaptive_threshold(img)
    img = scale_image(img)

    return img
```

```python
def CTCLoss(targets, outputs):
    batch_size = tf.cast(tf.shape(targets)[0], dtype="int64")
    output_seq_length = tf.cast(tf.shape(outputs)[1], dtype="int64")
    target_seq_length = tf.cast(tf.shape(targets)[1], dtype="int64")

    output_seq_length = output_seq_length * tf.ones(shape=(batch_size, 1), dtype="int64")
    target_seq_length = target_seq_length * tf.ones(shape=(batch_size, 1), dtype="int64")

    calculated_loss = K.ctc_batch_cost(targets, outputs, output_seq_length, target_seq_length)
    return calculated_loss
```

```python
def decode_label(num, alphabet):
    return ''.join(alphabet[ch] for ch in num if ch != len(alphabet))

def decode_predictions(nums):
    input_length = np.ones(nums.shape[0]) * nums.shape[1]
    values = get_value(
        ctc_decode(nums, input_length=input_length, greedy=True)[0][0]
    )

    texts = [decode_label(value[value >= 0], alphabet) for value in values]
    return texts


def encode_labels(texts):
    def _decode_label(label, alphabet):
        return np.array([alphabet.find(ch) for ch in label])

    # Создание алфавита из текстов
    alphabet = ''.join(sorted(set(''.join(texts))))

    max_len = max(len(text) for text in texts)
    nums = np.full((len(texts), max_len), fill_value=len(alphabet), dtype='int64')

    for i, text in enumerate(texts):
        nums[i, :len(text)] = _decode_label(text, alphabet)

    return nums, alphabet
```

```python
def CTCLoss(targets, outputs):
    batch_size = tf.cast(tf.shape(targets)[0], dtype="int64")
    output_seq_length = tf.cast(tf.shape(outputs)[1], dtype="int64")
    target_seq_length = tf.cast(tf.shape(targets)[1], dtype="int64")

    output_seq_length = output_seq_length * tf.ones(shape=(batch_size, 1), dtype="int64")
    target_seq_length = target_seq_length * tf.ones(shape=(batch_size, 1), dtype="int64")

    calculated_loss = K.ctc_batch_cost(targets, outputs, output_seq_length, target_seq_length)
    return calculated_loss
```

```python
import tensorflow as tf
from tensorflow.keras import backend as K

class CharacterErrorRate(tf.keras.metrics.Metric):
    def __init__(self, pad_token, name="cer", **kwargs):
        super().__init__(name=name, **kwargs)
        self.pad_token = pad_token

        self.total_edits = self.add_weight(name="total_edits", initializer="zeros")
        self.total_chars = self.add_weight(name="total_chars", initializer="zeros")

    def update_state(self, y_true, y_pred, sample_weight=None):
        batch_size = tf.shape(y_pred)[0]
        time_steps = tf.shape(y_pred)[1]

        input_lengths = tf.fill([batch_size], time_steps)

        decoded, _ = K.ctc_decode(y_pred, input_lengths, greedy=True)
        decoded_dense = decoded[0]

        mask = tf.not_equal(y_true, self.pad_token)

        label_lengths = tf.reduce_sum(tf.cast(mask, tf.int32), axis=1)

        sparse_true = K.ctc_label_dense_to_sparse(y_true, label_lengths)

        pred_mask = tf.not_equal(decoded_dense, -1)
        pred_lengths = tf.reduce_sum(tf.cast(pred_mask, tf.int32), axis=1)

        sparse_pred = K.ctc_label_dense_to_sparse(decoded_dense, pred_lengths)

        distances = tf.edit_distance(sparse_pred, sparse_true, normalize=False)

        self.total_edits.assign_add(tf.reduce_sum(distances))

        self.total_chars.assign_add(tf.reduce_sum(tf.cast(label_lengths, tf.float32)))

    def result(self):
        return tf.math.divide_no_nan(self.total_edits, self.total_chars)

    def reset_state(self):
        self.total_edits.assign(0.0)
        self.total_chars.assign(0.0)
```

```python
from tensorflow.keras.applications import VGG16
from tensorflow.keras.layers import Input, Reshape, Dense, Dropout, Bidirectional, LSTM, LeakyReLU
from tensorflow.keras.models import Model

def create_model_with_transfer_learning(num_classes, input_shape=(200, 64, 1)):
    full_vgg = VGG16(weights='imagenet', include_top=False, input_shape=(input_shape[0], input_shape[1], 3))

    base_model = Model(inputs=full_vgg.input, outputs=full_vgg.get_layer('block3_pool').output)

    input_img = Input(shape=input_shape, name='image_input')
    x = Conv2D(3, (1, 1), padding='same', name='proj_1to3')(input_img)
    x = base_model(x, training=False)

    base_model.trainable = False

    x = Conv2D(256, (3, 3), padding='same')(x)
    x = BatchNormalization()(x)
    x = LeakyReLU(0.01)(x)

    x = Conv2D(256, (3, 3), padding='same')(x)
    x = BatchNormalization()(x)
    x = LeakyReLU(0.01)(x)

    print("Shape before reshape:", x.shape)
    x = Reshape((x.shape[1], -1))(x)

    base_model.trainable = False

    x = Dense(256)(x)
    x = LeakyReLU(0.01)(x)
    x = Dropout(0.5)(x)

    x = Bidirectional(LSTM(256, return_sequences=True))(x)
    x = Bidirectional(LSTM(256, return_sequences=True))(x)

    x = Dense(256)(x)
    x = LeakyReLU(0.01)(x)
    x = Dropout(0.5)(x)

    x = Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=input_img, outputs=x)
    model.compile(optimizer=Nadam(learning_rate=0.001, clipnorm=1.0), loss=CTCLoss, metrics=[CharacterErrorRate(pad_token=num_classes - 1)])
    model.summary()
    return model
```

```python
def load_dataset(word_dict):
    image_base_folder = f'/kaggle/input/iam-handwriting-word-database/iam_words/words'
    train_X, train_y = [], []
    total_count = 0
    folder_counts = {}
    max_per_folder = 200
    max_total = 15000

    for root, _, files in os.walk(image_base_folder):
        if total_count >= max_total:
            break

        folder_name = os.path.basename(root)
        folder_counts[folder_name] = 0

        for image_file in files:
            if total_count >= max_total:
                break

            if folder_counts[folder_name] >= max_per_folder:
                break

            if image_file.endswith('.jpg') or image_file.endswith('.png'):
                image_name = image_file.rsplit('.', 1)[0]

                if image_name not in word_dict:
                    continue

                label = word_dict[image_name]
                if label == 'Ps':
                  continue

                if not any(c not in string.ascii_letters + ',.\' ' for c in label):
                    image_path = os.path.join(root, image_file)
                    if os.path.exists(image_path):
                        try:
                            img_array = cv2.imread(image_path, cv2.IMREAD_COLOR)
                            if img_array is None:
                                raise ValueError("Failed to read image with OpenCV")

                            if len(label) <= 12:
                                img_array = cv2.cvtColor(img_array, cv2.COLOR_BGR2RGB)
                                train_X.append(transform_image(img_array))
                                train_y.append(label)
                                folder_counts[folder_name] += 1
                                total_count += 1
                        except Exception as e:
                            print(f"Failed to load image: {image_path}, error: {e}")
                    else:
                        print(f"File does not exist: {image_path}")

    return np.array(train_X), np.array(train_y)
```

```python

train_X, train_y = load_dataset(word_dict)
train_y, charset = encode_labels(train_y)
```

```python

train_X, test_X, train_y, test_y = train_test_split(train_X, train_y, test_size=0.2, random_state=42)

train_X, val_X, train_y, val_y = train_test_split(train_X, train_y, test_size=0.2, random_state=42)

print("Обработка данных завершена")

num_classes = len(charset) + 1
print(charset)
print(f"Размерность y_true: {train_y.shape}")
print(num_classes)
input_dimensions = (200, 64, 1)
```

```python
model = create_model_with_transfer_learning(num_classes, input_shape=(200, 64, 1))
```

```python
early_stopping = EarlyStopping(
    monitor='val_cer',
    patience=10,
    restore_best_weights=True,
    mode='min'
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    min_lr=1e-5,
    patience=4,
    mode='min'
)

history = model.fit(
    train_X, train_y,
    validation_data=(val_X, val_y),
    epochs=60,
    batch_size=64,
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)
```
